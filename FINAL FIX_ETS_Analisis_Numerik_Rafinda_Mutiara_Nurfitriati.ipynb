## 1A. Teori dan Rumus Ringkas

**Massa vs berat.** Massa $m$ (kg, ton) berbeda dengan berat/gaya $W=mg$ (N, kN). Gaya apung dinyatakan dalam kN dan ton-gaya (1 ton-gaya = berat 1 ton massa = 9,81 kN).

**Volume benda ekstrusi.** $V=A_{profil}\,B$. Segitiga: $A=\tfrac12\,\text{alas}\times\text{tinggi}$, titik berat = rata-rata koordinat titik sudut. Seperempat elips: $A=\tfrac14\pi ab$, titik berat berjarak $\dfrac{4b}{3\pi}$ dari sumbu horizontal pusat.

**Archimedes.** Massa air terdesak $=\rho V_{tenggelam}$; gaya apung $F_b=\rho g V_{tenggelam}$. Dengan $\rho=1$ ton/m³, massa air terdesak (ton) bernilai sama dengan volume (m³).

**Panjang garis air** pada tinggi $y$ dari lantai: $L(y)=6+y+\sqrt{1-\left(\tfrac{y-2}{2}\right)^2}$; luas profil tenggelam $A(x)=\int_0^x L(y)\,dy$; $F_b(x)=\rho g B A(x)$. Kemiringan kurva $dF_b/dx=\rho g B\,L(x)$ naik dari $3\cdot6=18$ ton-gaya/m (176 kN/m) di lantai ke $3\cdot9=27$ ton-gaya/m (265 kN/m) di dek.

**Center of buoyancy:** $KB=\dfrac{\int_0^x y\,L(y)\,dy}{\int_0^x L(y)\,dy}$.

**Massa baja:** $m=\sigma A_{baja}$, $\sigma=160$ kg/m². Panjang busur buritan $s=\int_0^{\pi/2}\sqrt{a^2\cos^2t+b^2\sin^2t}\,dt$.

**Kesetimbangan apung:** $\rho V(x)=m_{total}$ (dalam kg). Bila $\rho V_{total}<m_{total}$ tidak ada akar pada $[0,2]$ m: kapal tenggelam.

**Center of mass:** $y_{CM}=\dfrac{\sum m_i y_i}{\sum m_i}$.

**Metode numerik:** kuadratur adaptif `quad` (utama), Simpson dan trapesium (pembanding galat), pencarian akar bisection (konvergen linear, pasti), Newton-Raphson (konvergen kuadratik) dan `brentq` (pembanding).

## 3. Model Geometri

Karena lebar $B$ konstan, semua besaran dihitung dari **panjang garis air** $L(y)$ (panjang kapal yang terpotong bidang horizontal setinggi $y$ dari lantai):

$$L(y) = L_{alas} + y\,\frac{L_{haluan}}{H} + \underbrace{a\sqrt{1-\Big(\frac{H-y}{b}\Big)^2}}_{\text{buritan elips}} = 6 + y + \tfrac12\sqrt{4y-y^2}$$

Luas bidang air $A_{wp}(y)=B\,L(y)$, sehingga

$$V_{tenggelam}(x)=\int_0^x B\,L(y)\,dy,\qquad F_{apung}(x)=\rho\,g\,V_{tenggelam}(x).$$

```python
def x_buritan(y):
    """Jarak horizontal buritan elips dari awal buritan pada ketinggian y (m)."""
    u = H - np.clip(y, 0.0, H)
    return A_EL * np.sqrt(np.maximum(1.0 - (u / B_EL)**2, 0.0))

def L_wl(y):
    """Panjang garis air (m) pada ketinggian y dari lantai kapal."""
    y = np.clip(y, 0.0, H)
    return L_ALAS + y * L_HALUAN / H + x_buritan(y)

def A_sub_eksak(d):
    """Luas profil (sisi) yang tenggelam sampai tinggi d, rumus eksak (m^2)."""
    d = np.clip(d, 0.0, H)
    s0 = np.clip((H - d) / B_EL, 0.0, 1.0)
    stern = A_EL * B_EL * (np.pi / 4 - 0.5 * (s0 * np.sqrt(1 - s0**2) + np.arcsin(s0)))
    return L_ALAS * d + L_HALUAN / (2 * H) * d**2 + stern

def V_sub(d):
    """Volume tenggelam (m^3) sampai tinggi d (eksak)."""
    return B * A_sub_eksak(d)

def gambar_kapal(ax, d=None, beban_h=None):
    """Profil kapal (tampak samping). d = garis air, beban_h = tinggi muatan (m)."""
    t = np.linspace(0, np.pi / 2, 80)
    xs = [-L_HALUAN, 0.0, L_ALAS] + list(L_ALAS + A_EL * np.sin(t)) + [L_ALAS + A_EL, -L_HALUAN]
    ys = [H, 0.0, 0.0] + list(H - B_EL * np.cos(t)) + [H, H]
    ax.fill(xs, ys, color="lightsteelblue", ec="navy", lw=2, zorder=2)
    if beban_h is not None:
        ax.add_patch(plt.Rectangle((L_ALAS / 2 - 1.5, H), 4.0, beban_h, color="mistyrose", ec="firebrick", lw=2, zorder=3))
        ax.text(L_ALAS / 2 + 0.5, H + beban_h / 2, "muatan", ha="center", va="center", fontsize=9)
    if d is not None:
        ax.axhline(d, color="tab:blue", ls="--", lw=1.5, zorder=4)
        ax.fill([-L_HALUAN - 1, L_ALAS + A_EL + 1, L_ALAS + A_EL + 1, -L_HALUAN - 1], [0, 0, d, d], color="tab:blue", alpha=0.12, zorder=1)
    ax.set_aspect("equal"); ax.set_xlabel("x memanjang (m)"); ax.set_ylabel("tinggi dari lantai (m)")

fig, ax = plt.subplots(figsize=(10, 4))
gambar_kapal(ax, beban_h=H_BEBAN_CM / 100)
ax.set_title("Profil kapal (tampak samping) dan letak muatan di dek"); plt.tight_layout(); plt.show()

print(f"L(0) = {L_wl(0):.2f} m (dasar datar)   L(H) = {L_wl(H):.2f} m (garis dek = 2 + 6 + 1)")
```

```python
from matplotlib.patches import Polygon, Rectangle
# Tiga proyeksi berlabel: tampak samping, tampak atas, potongan melintang badan tengah
def poligon_profil(n=120):
    t = np.linspace(0, np.pi/2, n)
    arc = np.column_stack([L_ALAS + A_EL*np.sin(t), H - B_EL*np.cos(t)])
    return np.vstack([[-L_HALUAN, H], [0, 0], [L_ALAS, 0], arc[1:], [-L_HALUAN, H]])

def panah(ax, p, q, teks, dx=0.0, dz=0.0):
    ax.annotate("", xy=q, xytext=p, arrowprops=dict(arrowstyle="<->", lw=1.1, color="k"))
    ax.text((p[0]+q[0])/2 + dx, (p[1]+q[1])/2 + dz, teks, ha="center", va="center", fontsize=9,
            bbox=dict(fc="white", ec="none", pad=1.0))

fig = plt.figure(figsize=(11, 7.2))
gs = fig.add_gridspec(2, 2, height_ratios=[1.35, 1], hspace=0.35, wspace=0.25)

ax = fig.add_subplot(gs[0, :])
ax.add_patch(Polygon(poligon_profil(), closed=True, fc="#dfe9f5", ec="#1f4e8c", lw=2))
for s in (0, L_ALAS):
    ax.plot([s, s], [0, H], "--", color="#1f4e8c", lw=1)
ax.plot([-L_HALUAN, L_ALAS + A_EL], [H, H], color="#1f4e8c", lw=2.5)
tt = np.linspace(np.pi, 2*np.pi, 200)                      # setengah elips bawah sebagai referensi
ax.plot(L_ALAS + A_EL*np.cos(tt), H + B_EL*np.sin(tt), ":", color="gray", lw=0.8)
ax.plot(L_ALAS, H, "k.", ms=4); ax.text(L_ALAS + 0.05, H + 0.08, "pusat elips (6, 2)", fontsize=8)
panah(ax, (-L_HALUAN, H + 0.35), (0, H + 0.35), "2 m", dz=0.18)
panah(ax, (0, H + 0.35), (L_ALAS, H + 0.35), "6 m", dz=0.18)
panah(ax, (L_ALAS, H + 0.35), (L_ALAS + A_EL, H + 0.35), "1 m", dz=0.18)
panah(ax, (L_ALAS + A_EL + 0.35, 0), (L_ALAS + A_EL + 0.35, H), "2 m", dx=0.18)
panah(ax, (-L_HALUAN - 0.35, 0), (-L_HALUAN - 0.35, H), "2 m", dx=-0.3)
ax.text(-0.95, 0.55, "45°", fontsize=10, color="#b00020")
ax.text(-0.75, 1.0, "haluan\n(prisma segitiga)", ha="center", fontsize=9)
ax.text(3.0, 1.0, "badan tengah (balok)\n6 m × 3 m × 2 m", ha="center", fontsize=10)
ax.text(6.55, 1.25, "buritan\n(¼ elips)", ha="center", fontsize=9)
ax.text(-0.35, -0.35, "lantai/lunas y = 0", fontsize=8)
ax.set_aspect("equal"); ax.set_xlim(-3.2, 8.4); ax.set_ylim(-0.7, 3.0)
ax.set_xlabel("posisi memanjang (m)"); ax.set_ylabel("tinggi dari lantai (m)")
ax.set_title("Tampak samping (profil), lebar kapal B = 3 m tegak lurus kertas")

ax2 = fig.add_subplot(gs[1, 0])
ax2.add_patch(Rectangle((-L_HALUAN, 0), L_HALUAN + L_ALAS + A_EL, B, fc="#e8efe0", ec="#3b6b1f", lw=2))
for s in (0, L_ALAS):
    ax2.plot([s, s], [0, B], "--", color="#3b6b1f", lw=1)
panah(ax2, (-L_HALUAN - 0.3, 0), (-L_HALUAN - 0.3, B), "3 m", dx=-0.35)
panah(ax2, (-L_HALUAN, -0.45), (L_ALAS + A_EL, -0.45), "9 m (denah dek)", dz=-0.3)
ax2.set_aspect("equal"); ax2.set_xlim(-3.3, 8); ax2.set_ylim(-1.3, 3.5)
ax2.set_title("Tampak atas (dek)"); ax2.set_xlabel("posisi memanjang (m)"); ax2.set_ylabel("lebar (m)")

ax3 = fig.add_subplot(gs[1, 1])
ax3.add_patch(Rectangle((0, 0), B, H, fc="#f3e3d3", ec="#8c4a1f", lw=2))
panah(ax3, (0, -0.35), (B, -0.35), "3 m", dz=-0.25)
panah(ax3, (B + 0.3, 0), (B + 0.3, H), "2 m", dx=0.3)
ax3.set_aspect("equal"); ax3.set_xlim(-0.8, 4.4); ax3.set_ylim(-1.0, 2.8)
ax3.set_title("Potongan melintang badan tengah"); ax3.set_xlabel("lebar (m)"); ax3.set_ylabel("tinggi (m)")
plt.show()
```

## 4. Soal 1: Volume Total Kapal dan Gaya Apung Maksimum

**Langkah:** volume = luas profil dikali lebar $B$, dijumlahkan untuk tiga bagian.

- Haluan (prisma segitiga): $V_1=\tfrac12\,L_{haluan}\,H\,B$
- Badan tengah (balok): $V_2=L_{alas}\,H\,B$
- Buritan (¼ elips): $V_3=\tfrac14\pi\,a\,b\,B$

Gaya apung maksimum terjadi saat kapal tercelup penuh: $F_{max}=\rho\,g\,V_{total}$.

```python
V_haluan = 0.5 * L_HALUAN * H * B
V_tengah = L_ALAS * H * B
V_buritan = 0.25 * np.pi * A_EL * B_EL * B
V_total = V_haluan + V_tengah + V_buritan

tab_v = pd.DataFrame({
    "Bagian": ["Haluan (prisma segitiga)", "Badan tengah (balok)", "Buritan (1/4 elips)", "TOTAL"],
    "Rumus": ["1/2 x 2 x 2 x 3", "6 x 2 x 3", "1/4 x pi x 1 x 2 x 3", ""],
    "Volume (m3)": [V_haluan, V_tengah, V_buritan, V_total],
})
print(tab_v.to_string(index=False))

# Pemeriksaan numerik: V = B * integral L(y) dy  (kuadratur adaptif)
V_num, galat_num = quad(lambda y: B * L_wl(y), 0, H)
print(f"\nCek integral numerik  : {V_num:.8f} m^3  (galat taksiran {galat_num:.1e})")
print(f"Selisih terhadap eksak: {abs(V_num - V_total):.2e} m^3")

m_air = RHO * V_total                 # massa air yang dipindahkan (kg)
F_max = RHO * G * V_total             # newton
print(f"\nMassa air terpindah = rho x V = {RHO:.0f} x {V_total:.4f} = {m_air:,.1f} kg = {m_air/1000:.3f} ton")
print(f"Gaya apung maksimum = rho g V = {RHO:.0f} x {G} x {V_total:.4f} = {F_max:,.1f} N = {F_max/1000:.2f} kN")
print(f"                    = {m_air:,.1f} kgf = {m_air/1000:.2f} ton-gaya")
```

### 4.1 Pemeriksaan Galat Integrasi Numerik

Bagian buritan memuat $\tfrac12\sqrt{4y-y^2}$ yang turunannya tak berhingga di $y=0$. Simpson biasa menjadi **lambat konvergen** untuk fungsi seperti ini, sedangkan kuadratur adaptif (`quad`) menanganinya dengan baik. Tabel membandingkan keduanya pada luas buritan, yang nilai eksaknya $\tfrac{\pi}{4}ab=\tfrac{\pi}{2}$.

```python
eksak = np.pi * A_EL * B_EL / 4
baris = []
for n_ in [10, 20, 40, 80, 160, 320, 640]:
    yy = np.linspace(0, H, n_ + 1)
    s = simpson(x_buritan(yy), x=yy)
    baris.append((n_, s, abs(s - eksak)))
q, _ = quad(x_buritan, 0, H)

pd.options.display.float_format = "{:.6e}".format
print(pd.DataFrame(baris, columns=["n segmen", "Simpson", "|galat|"]).to_string(index=False))
pd.options.display.float_format = "{:,.4f}".format
print(f"\nNilai eksak = pi/2 = {eksak:.10f}")
print(f"quad        = {q:.10f}  (galat {abs(q - eksak):.1e})")
```

## 5. Soal 2: Kurva Tinggi Tenggelam ($x$) terhadap Gaya Apung ($y$)

**Langkah:** untuk setiap $x$ dari 0 sampai 2 m, hitung $V_{tenggelam}(x)=B\int_0^x L(y)\,dy$ secara numerik (`quad`), lalu $F(x)=\rho\,g\,V_{tenggelam}(x)$. Hasilnya dibandingkan dengan rumus eksak.

```python
x_tenggelam = np.linspace(0, H, 201)                                   # tinggi tenggelam (m)
V_num_x = np.array([B * quad(L_wl, 0, xi)[0] for xi in x_tenggelam])   # numerik
V_eks_x = V_sub(x_tenggelam)                                           # eksak
F_x = RHO * G * V_num_x                                                # gaya apung (N)

print(f"Galat maksimum numerik vs eksak : {np.max(np.abs(V_num_x - V_eks_x)):.2e} m^3")
print(f"F(x = 2 m) = {F_x[-1]/1000:.2f} kN  (harus sama dengan F maksimum = {F_max/1000:.2f} kN)")

titik_tabel = np.array([0, 0.25, 0.5, 0.75, 1.0, 1.25, 1.5, 1.75, 2.0])
tab_F = pd.DataFrame({
    "x tenggelam (m)": titik_tabel,
    "x (cm)": titik_tabel * 100,
    "V tenggelam (m3)": V_sub(titik_tabel),
    "F apung (kN)": RHO * G * V_sub(titik_tabel) / 1000,
    "F apung (ton-gaya)": RHO * V_sub(titik_tabel) / 1000,
})
print()
print(tab_F.to_string(index=False))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x_tenggelam, F_x / 1000, color="tab:blue", lw=2.2, label="F(x) numerik (quad)")
ax.plot(titik_tabel, RHO * G * V_sub(titik_tabel) / 1000, "o", color="tab:red", ms=5, label="titik tabel (rumus eksak)")
ax.set_xlabel("Ketinggian tenggelam x dari lantai kapal (m)")
ax.set_ylabel("Gaya apung y (kN)")
ax.set_title("Kurva ketinggian tenggelam (x) vs gaya apung (y)")
ax.legend(); plt.tight_layout(); plt.show()
```

Kurva naik terus dan **kemiringannya bertambah** seiring $x$ (melengkung ke atas), karena $\frac{dF}{dx}=\rho\,g\,B\,L(x)$ dan garis air $L(x)$ makin panjang dari 6 m di lantai sampai 9 m di dek. Pada $x = 2$ m kurva berhenti di $F_{max}$. Kemiringan $dF/dx=\rho g B\,L(x)$ naik dari $3\times6=18$ ton-gaya/m (176 kN/m) di lantai ke $3\times9=27$ ton-gaya/m (265 kN/m) di dek.

**Ilustrasi garis air** pada beberapa tinggi tenggelam:

## 6. Soal 3: Tinggi *Center of Buoyancy* dari Lantai Kapal

*Center of buoyancy* adalah titik berat volume air yang dipindahkan. Karena lebar konstan, cukup dihitung dari profil:

$$KB(x)=\frac{\displaystyle\int_0^x y\,L(y)\,dy}{\displaystyle\int_0^x L(y)\,dy}$$

Untuk kondisi tercelup penuh ($x=H$), hasilnya dicek dengan titik berat gabungan tiga bangun:

| Bagian | Luas profil (m²) | Tinggi titik berat (m) |
|---|---|---|
| Badan tengah (6×2) | 12 | 1 |
| Haluan (segitiga siku-siku) | 2 | 4/3 |
| Buritan (¼ elips, pusat di dek) | $\pi a b/4=\pi/2$ | $H-\dfrac{4b}{3\pi}$ |

```python
def KB(d):
    """Tinggi center of buoyancy dari lantai (m) untuk tinggi tenggelam d."""
    pembilang, _ = quad(lambda y: y * L_wl(y), 0, d)
    penyebut, _ = quad(L_wl, 0, d)
    return pembilang / penyebut

# --- Cek eksak: kondisi tercelup penuh (x = H) ---
A_tengah, y_tengah = L_ALAS * H, H / 2
A_haluan = 0.5 * L_HALUAN * H
y_haluan = (0 + H + H) / 3                          # titik berat segitiga: (0 + 2 + 2)/3
A_buritan = np.pi * A_EL * B_EL / 4
y_buritan = H - 4 * B_EL / (3 * np.pi)

A_tot = A_tengah + A_haluan + A_buritan             # luas satu profil sisi
KB_gabungan = (A_tengah * y_tengah + A_haluan * y_haluan + A_buritan * y_buritan) / A_tot
KB_penuh = KB(H)

tab_kb = pd.DataFrame({
    "Bagian": ["Badan tengah", "Haluan", "Buritan", "TOTAL"],
    "Luas (m2)": [A_tengah, A_haluan, A_buritan, A_tot],
    "y titik berat (m)": [y_tengah, y_haluan, y_buritan, KB_gabungan],
    "Luas x y (m3)": [A_tengah * y_tengah, A_haluan * y_haluan, A_buritan * y_buritan, KB_gabungan * A_tot],
})
print(tab_kb.to_string(index=False))
print(f"\nKB (titik berat gabungan, eksak) = {KB_gabungan:.5f} m")
print(f"KB (integral numerik quad)       = {KB_penuh:.5f} m")
print(f"\n==> Tinggi center of buoyancy (tercelup penuh) = {KB_penuh*100:.1f} cm  ~ {round(KB_penuh*100)} cm dari lantai kapal")
```

```python
# KB sebagai fungsi tinggi tenggelam (berguna bila kapal tidak tercelup penuh)
d_kb = np.linspace(0.02, H, 100)
kb_kurva = np.array([KB(di) for di in d_kb])

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(d_kb, kb_kurva * 100, color="tab:green", lw=2.2)
ax.scatter([H], [KB_penuh * 100], color="red", zorder=5)
ax.annotate(f"KB = {KB_penuh*100:.0f} cm (tercelup penuh)", (H, KB_penuh * 100), textcoords="offset points", xytext=(-190, -22))
ax.set_xlabel("Ketinggian tenggelam x (m)"); ax.set_ylabel("KB dari lantai (cm)")
ax.set_title("Tinggi center of buoyancy terhadap ketinggian tenggelam"); plt.tight_layout(); plt.show()
```

## 7. Soal 4: Massa Total Kapal (Pelat Baja 160 kg/m²)

**Langkah:** hitung luas seluruh permukaan pelat (termasuk dek atas), lalu kalikan 160 kg/m².

| Pelat | Luas |
|---|---|
| Dek atas | $(L_{haluan}+L_{alas}+a)\times B$ |
| Dasar datar | $L_{alas}\times B$ |
| Dua sisi | $2\times$ luas profil |
| Bidang miring haluan | $\sqrt{L_{haluan}^2+H^2}\times B$ |
| Selongsong buritan | panjang busur ¼ elips $\times B$ |

Panjang busur ¼ elips tidak punya rumus sederhana (integral eliptik), jadi dihitung numerik. Dengan $x=a\sin t,\ y=H-b\cos t$ untuk $t\in[0,\pi/2]$:

$$s=\int_0^{\pi/2}\sqrt{a^2\cos^2 t+b^2\sin^2 t}\;dt$$

```python
def ds_busur(t):
    return np.sqrt((A_EL * np.cos(t))**2 + (B_EL * np.sin(t))**2)

s_busur, galat_s = quad(ds_busur, 0, np.pi / 2)
# Pembanding: aproksimasi Ramanujan untuk keliling elips, dibagi 4
s_ramanujan = np.pi * (3 * (A_EL + B_EL) - np.sqrt((3 * A_EL + B_EL) * (A_EL + 3 * B_EL))) / 4
print(f"Panjang busur 1/4 elips (quad)       : {s_busur:.6f} m  (galat taksiran {galat_s:.1e})")
print(f"Pembanding aproksimasi Ramanujan/4   : {s_ramanujan:.6f} m  (selisih {abs(s_busur - s_ramanujan):.1e})")

panjang_dek = L_HALUAN + L_ALAS + A_EL
S_dek = panjang_dek * B
S_dasar = L_ALAS * B
S_sisi = 2 * A_tot
S_miring = np.hypot(L_HALUAN, H) * B
S_buritan = s_busur * B
S_total = S_dek + S_dasar + S_sisi + S_miring + S_buritan

tab_S = pd.DataFrame({
    "Pelat": ["Dek atas", "Dasar datar", "Dua sisi", "Bidang miring haluan", "Selongsong buritan", "TOTAL"],
    "Luas (m2)": [S_dek, S_dasar, S_sisi, S_miring, S_buritan, S_total],
})
tab_S["Massa baja (kg)"] = tab_S["Luas (m2)"] * BAJA_AREAL
print()
print(tab_S.to_string(index=False))

m_baja = BAJA_AREAL * S_total
print(f"\nMassa kapal (baja) = {BAJA_AREAL:.0f} kg/m^2 x {S_total:.4f} m^2 = {m_baja:,.2f} kg = {m_baja/1000:.3f} ton")
print(f"Berat kapal        = {m_baja*G:,.1f} N = {m_baja*G/1000:.2f} kN")
```

## 8. Soal 5: Kedalaman Kapal yang Terendam Saat Memuat Muatan

**Langkah:**

1. Massa total $W = m_{baja} + m_{muatan}$.
2. Kapal terapung bila gaya apung = berat, yaitu $\rho\,V_{tenggelam}(x) = W$ (dalam kg).
3. Selesaikan $f(x)=\rho\,V_{tenggelam}(x)-W=0$ secara numerik (bisection, Newton-Raphson, dan `brentq` sebagai pembanding), dengan $f'(x)=\rho\,B\,L(x)$.
4. Syarat mengapung: $W \le \rho\,V_{total}$. Bila tidak terpenuhi, kapal tenggelam.

```python
def bisection(f, a, b, tol=1e-8, nmax=100):
    fa, fb = f(a), f(b)
    if fa * fb > 0:
        raise ValueError("Tidak ada akar pada interval.")
    log = []
    for k in range(1, nmax + 1):
        c = (a + b) / 2
        fc = f(c)
        log.append((k, a, b, c, fc))
        if abs(fc) < tol or (b - a) / 2 < tol:
            break
        if fa * fc < 0:
            b, fb = c, fc
        else:
            a, fa = c, fc
    return c, log

def newton(f, df, x0, tol=1e-10, nmax=50):
    log, xk = [], x0
    for k in range(1, nmax + 1):
        x_baru = xk - f(xk) / df(xk)
        log.append((k, xk, f(xk), x_baru))
        if abs(x_baru - xk) < tol:
            return x_baru, log
        xk = x_baru
    return xk, log

def kedalaman_terendam(W_kg, tampil=False):
    """Kedalaman terendam (m), atau None bila kapal tenggelam."""
    f = lambda d: RHO * V_sub(d) - W_kg
    if f(H) < 0:
        return None
    d_bis, log_b = bisection(f, 0.0, H)
    d_new, log_n = newton(f, lambda d: RHO * B * L_wl(d), 1.0)
    d_brq = brentq(f, 0.0, H, xtol=1e-12)
    if tampil:
        print("Bisection (beberapa iterasi pertama):")
        print(pd.DataFrame(log_b[:8], columns=["iterasi", "a", "b", "c", "f(c) kg"]).to_string(index=False))
        print("\nNewton-Raphson:")
        print(pd.DataFrame(log_n, columns=["iterasi", "x_k", "f(x_k) kg", "x_k+1"]).to_string(index=False))
        print(f"\nBisection : {d_bis:.6f} m ({len(log_b)} iterasi)")
        print(f"Newton    : {d_new:.6f} m ({len(log_n)} iterasi)")
        print(f"brentq    : {d_brq:.6f} m")
    return d_brq

# ============ Skenario A: sesuai teks soal (muatan dalam TON) ============
m_muatan_A = MUATAN_TON * 1000.0                 # kg
W_A = m_baja + m_muatan_A
print("=== SKENARIO A: muatan sesuai teks soal (ton) ===")
print(f"Massa muatan      = {MUATAN_TON:,.0f} ton = {m_muatan_A:,.0f} kg")
print(f"Massa total       = {m_baja:,.1f} + {m_muatan_A:,.0f} = {W_A:,.1f} kg = {W_A/1000:,.2f} ton")
print(f"Daya apung maks   = {RHO*V_total:,.1f} kg = {RHO*V_total/1000:.2f} ton")
d_A = kedalaman_terendam(W_A)
if d_A is None:
    print(f"\n-> Massa total ({W_A/1000:,.1f} ton) melebihi daya apung maksimum ({RHO*V_total/1000:.2f} ton).")
    print(f"-> KAPAL TENGGELAM: seluruh tinggi kapal ({H*100:.0f} cm) terendam dan masih kurang.")
    print(f"   Volume yang dibutuhkan = {W_A/RHO:,.1f} m^3, sedangkan volume kapal hanya {V_total:.2f} m^3.")
else:
    print(f"\n-> Kedalaman terendam = {d_A*100:.1f} cm")

m_muatan_maks = RHO * V_total - m_baja
print(f"\nMuatan maksimum yang masih bisa dibawa (garis air tepat di dek) = {m_muatan_maks:,.1f} kg = {m_muatan_maks/1000:.2f} ton")
```

**Catatan penting.** Dengan teks soal apa adanya (**1200 ton**), muatan jauh melebihi daya apung maksimum kapal ($\approx 46{,}7$ ton), sehingga kapal tenggelam dan tidak ada kedalaman terendam yang bisa dicari. Karena itu saya juga menghitung **skenario B** sebagai pembanding, yaitu bila angkanya dimaksudkan dalam **kg** (12 × 100 = 1200 kg), supaya proses pencarian akarnya dapat ditunjukkan.

```python
# ============ Skenario B: bila muatan dimaksudkan dalam KG ============
m_muatan_B = n3 * 100.0                          # kg
W_B = m_baja + m_muatan_B
print("=== SKENARIO B: muatan dalam kg (pembanding) ===")
print(f"Massa muatan  = {n3} x 100 = {m_muatan_B:,.0f} kg")
print(f"Massa total   = {m_baja:,.1f} + {m_muatan_B:,.0f} = {W_B:,.1f} kg\n")
d_B = kedalaman_terendam(W_B, tampil=True)
print(f"\n-> Kedalaman terendam (skenario B) = {d_B:.4f} m = {d_B*100:.1f} cm  ~ {round(d_B*100)} cm")
print(f"-> Freeboard (jarak garis air ke dek) = {(H - d_B)*100:.1f} cm")
print(f"\nCek: rho x V_tenggelam({d_B:.4f}) = {RHO*V_sub(d_B):,.2f} kg  vs  W = {W_B:,.2f} kg")
```

## 9. Soal 6: Kurva Tinggi dari Lantai ($x$) terhadap Massa Akumulatif ($y$)

$h_{beban}=200-\text{NPM}=200-12=188$ cm, dan massa muatan terdistribusi rata secara vertikal dari dek ($x=2$ m) sampai $x=2+h_{beban}$.

**Langkah:** massa akumulatif $m(x)$ = massa semua bagian yang berada di bawah ketinggian $x$. Kurva pertama menggambarkan **muatan saja** ($m_{muatan}(x)=\lambda\,(x-2)$ dengan $\lambda=m_{muatan}/h_{beban}$, linear), sedangkan dua kurva berikutnya menggambarkan **seluruh sistem** (cangkang baja + muatan) dengan rincian berikut:

- Pelat dasar (18 m²): massa melompat di $x=0$.
- Dua sisi: $2\,\rho_{baja}\,A_{sub}(x)$, dengan $A_{sub}(x)=\int_0^x L(y)\,dy$.
- Bidang miring haluan: $\rho_{baja}\,B\,\dfrac{\sqrt{L_{haluan}^2+H^2}}{H}\,x$ (panjang miring bertambah merata terhadap tinggi).
- Selongsong buritan: $\rho_{baja}\,B\,s(x)$, dengan $s(x)=\int_0^{t(x)}\sqrt{a^2\cos^2\tau+b^2\sin^2\tau}\,d\tau$ dan $\cos t(x)=1-x/b$.
- Dek atas (27 m²): massa melompat di $x=2$ m.
- Muatan: naik linear dari 0 sampai $m_{muatan}$ antara $x=2$ dan $x=2+h_{beban}$.

```python
h_beban = H_BEBAN_CM / 100                      # tinggi muatan (m)
M_DEK = BAJA_AREAL * S_dek
M_DASAR = BAJA_AREAL * S_dasar

def m_baja_kumulatif(x):
    """Massa pelat baja (kg) yang berada di bawah ketinggian x (0 < x <= H); dek dimasukkan bila x > H."""
    if x <= 0:
        return 0.0
    xc = min(x, H)
    t_x = np.arccos(1.0 - xc / B_EL)
    m = M_DASAR
    m += BAJA_AREAL * 2 * float(A_sub_eksak(xc))
    m += BAJA_AREAL * B * (np.hypot(L_HALUAN, H) / H) * xc
    m += BAJA_AREAL * B * quad(ds_busur, 0, t_x)[0]
    if x > H + 1e-12:
        m += M_DEK
    return m

def m_kumulatif(x, m_muatan):
    """Massa akumulatif kapal + muatan (kg) sampai ketinggian x dari lantai."""
    return m_baja_kumulatif(x) + m_muatan * np.clip((x - H) / h_beban, 0.0, 1.0)

# Pemeriksaan: massa baja kumulatif di puncak harus sama dengan hasil soal 4
print(f"Massa baja kumulatif di x > 2 m : {m_baja_kumulatif(H + 1e-9):,.2f} kg")
print(f"Massa baja dari soal 4          : {m_baja:,.2f} kg   -> cocok: {np.isclose(m_baja_kumulatif(H + 1e-9), m_baja)}")

# Titik-titik kurva (lompatan di x = 0 dan x = 2 digambarkan tegak)
X_atas = H + h_beban
x_plot = np.concatenate([[0.0], np.linspace(1e-9, H, 200), [H + 1e-9], np.linspace(H + 1e-9, X_atas, 100)])

fig, axs = plt.subplots(2, 2, figsize=(13, 9))
lam_A = m_muatan_A / h_beban               # kg per meter tinggi (skenario A)
lam_B = m_muatan_B / h_beban               # kg per meter tinggi (skenario B)
z_c = np.linspace(0, X_atas + 0.4, 300)
titik_k = H + h_beban * np.linspace(0, 1, 11)

# (a) dan (b): massa akumulatif MUATAN saja (terdistribusi rata vertikal, jawaban langsung soal 6)
for ax, lam, skala, sat, judul in [(axs[0, 0], lam_A, 1000.0, "ton", f"Muatan saja, skenario A ({MUATAN_TON:,.0f} ton)"),
                                   (axs[0, 1], lam_B, 1.0, "kg", f"Muatan saja, skenario B ({m_muatan_B:,.0f} kg)")]:
    ax.plot(z_c, np.clip(z_c - H, 0, h_beban) * lam / skala, color="tab:orange", lw=2.2)
    ax.scatter(titik_k, np.clip(titik_k - H, 0, h_beban) * lam / skala, color="tab:red", zorder=3, s=18)
    ax.axvline(H, color="gray", ls="--", lw=1); ax.axvline(X_atas, color="gray", ls="--", lw=1)
    ax.set_xlabel("Ketinggian dari lantai kapal x (m)"); ax.set_ylabel(f"Massa akumulatif muatan y ({sat})")
    ax.set_title(judul)

# (c) dan (d): massa akumulatif seluruh sistem (cangkang baja + muatan)
for ax, mm, judul in [(axs[1, 0], m_muatan_A, f"Kapal + muatan {MUATAN_TON:,.0f} ton (skenario A)"),
                      (axs[1, 1], m_muatan_B, f"Kapal + muatan {m_muatan_B:,.0f} kg (skenario B)")]:
    y_plot = np.array([m_kumulatif(xi, mm) for xi in x_plot]) / 1000       # ton
    ax.plot(x_plot, y_plot, color="tab:purple", lw=2.2)
    ax.axvline(H, color="gray", ls="--", lw=1); ax.text(H, ax.get_ylim()[0], "  dek", va="bottom", fontsize=9)
    ax.set_xlabel("Ketinggian dari lantai kapal x (m)"); ax.set_ylabel("Massa akumulatif y (ton)")
    ax.set_title(judul)
plt.tight_layout(); plt.show()

tab_titik = pd.DataFrame({"x dari lantai (m)": titik_k,
                          "massa akumulatif muatan A (ton)": np.clip(titik_k - H, 0, h_beban) * lam_A / 1000,
                          "massa akumulatif muatan B (kg)": np.clip(titik_k - H, 0, h_beban) * lam_B})
print("Titik data kurva muatan saja:")
print(tab_titik.to_string(index=False))
print()

tab_m = pd.DataFrame({"x (m)": [0.5, 1.0, 1.5, 2.0, 2.5, 3.0, X_atas]})
tab_m["m kumulatif A (ton)"] = [m_kumulatif(xi + (1e-9 if xi >= H else 0), m_muatan_A) / 1000 for xi in tab_m["x (m)"]]
tab_m["m kumulatif B (kg)"] = [m_kumulatif(xi + (1e-9 if xi >= H else 0), m_muatan_B) for xi in tab_m["x (m)"]]
print(tab_m.to_string(index=False))
```

Kurva memiliki dua lompatan karena pelat dasar (di $x=0$) dan pelat dek (di $x=2$ m) adalah bidang datar yang massanya terkumpul pada satu ketinggian. Di antaranya kurva naik landai mengikuti sisi, bidang miring, dan busur buritan. Di atas dek kurva naik **linear** sampai $x=3{,}88$ m karena massa muatan terdistribusi rata secara vertikal.

## 10. Soal 7: Tinggi *Center of Mass* dari Lantai Kapal Beserta Muatan

**Langkah:**

1. Titik berat tiap pelat baja dikali luasnya: dek di $y=2$, dasar di $y=0$, bidang miring haluan di $y=1$, dua sisi di $y=KB_{penuh}$ (titik berat profil), dan selongsong buritan di titik berat busur $\bar y=\dfrac{1}{s}\displaystyle\int_0^{\pi/2}(H-b\cos t)\,ds$.
2. Muatan terdistribusi rata secara vertikal, sehingga $y_{muatan}=H+\tfrac12 h_{beban}$.
3. Titik berat gabungan:

$$y_{CM}=\frac{m_{baja}\,y_{baja}+m_{muatan}\,y_{muatan}}{m_{baja}+m_{muatan}}$$

4. **Pemeriksaan silang** dengan kurva soal 6: $\displaystyle y_{CM}=X-\frac{1}{M}\int_0^{X}m(x)\,dx$, karena $\int x\,dm = XM-\int m\,dx$.

```python
# Titik berat busur buritan (integral numerik)
y_busur = quad(lambda t: (H - B_EL * np.cos(t)) * ds_busur(t), 0, np.pi / 2)[0] / s_busur

y_s_dek, y_s_dasar, y_s_miring = H, 0.0, H / 2
y_s_sisi = KB_gabungan

tab_cm = pd.DataFrame({
    "Pelat": ["Dek atas", "Dasar datar", "Dua sisi", "Bidang miring haluan", "Selongsong buritan"],
    "Luas (m2)": [S_dek, S_dasar, S_sisi, S_miring, S_buritan],
    "y titik berat (m)": [y_s_dek, y_s_dasar, y_s_sisi, y_s_miring, y_busur],
})
tab_cm["Luas x y (m3)"] = tab_cm["Luas (m2)"] * tab_cm["y titik berat (m)"]
y_baja = tab_cm["Luas x y (m3)"].sum() / S_total
print(tab_cm.to_string(index=False))
print(f"\nTitik berat pelat baja : y_baja = {tab_cm['Luas x y (m3)'].sum():.4f} / {S_total:.4f} = {y_baja:.4f} m = {y_baja*100:.1f} cm")

y_muatan = H + h_beban / 2
print(f"Titik berat muatan     : y_muatan = {H} + {h_beban}/2 = {y_muatan:.3f} m = {y_muatan*100:.1f} cm")

def y_CM(m_muatan_kg):
    return (m_baja * y_baja + m_muatan_kg * y_muatan) / (m_baja + m_muatan_kg)

CM_A = y_CM(m_muatan_A)
CM_B = y_CM(m_muatan_B)

# Pemeriksaan silang lewat kurva massa akumulatif: y_CM = X - (1/M) * integral m(x) dx
def y_CM_dari_kurva(m_muatan):
    M_tot = m_baja + m_muatan
    I1 = quad(lambda x: m_kumulatif(x, m_muatan), 0, H, limit=200)[0]
    I2 = quad(lambda x: m_kumulatif(x, m_muatan), H + 1e-9, X_atas, limit=200)[0]
    return X_atas - (I1 + I2) / M_tot

print("\n--- Titik berat gabungan (kapal + muatan) ---")
print(f"Skenario A (muatan {MUATAN_TON:,.0f} ton): y_CM = {CM_A:.4f} m = {CM_A*100:.1f} cm  ~ {round(CM_A*100)} cm dari lantai")
print(f"   cek lewat kurva soal 6        : {y_CM_dari_kurva(m_muatan_A):.4f} m")
print(f"Skenario B (muatan {m_muatan_B:,.0f} kg) : y_CM = {CM_B:.4f} m = {CM_B*100:.1f} cm  ~ {round(CM_B*100)} cm dari lantai")
print(f"   cek lewat kurva soal 6        : {y_CM_dari_kurva(m_muatan_B):.4f} m")
print(f"Kapal kosong (tanpa muatan)    : y_CM = {y_baja*100:.1f} cm")
```

Hasil rumus langsung dan hasil dari kurva massa akumulatif sama, sehingga kedua cara saling memvalidasi. Pada skenario A titik berat gabungan sangat dekat dengan titik berat muatan karena muatan jauh lebih berat daripada kapal.
